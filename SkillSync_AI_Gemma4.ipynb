```python
import torch
from transformers import pipeline

if not torch.cuda.is_available():
    raise RuntimeError(
        "GPU not detected. Enable a GPU in Colab Runtime settings."
    )

MODEL_ID = "google/gemma-4-E2B-it"

gemma = pipeline(
    task="text-generation",
    model=MODEL_ID,
    device_map="auto",
    dtype="auto"
)

print("Gemma 4 loaded successfully!")
```

```python

import json
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output

# ----------------------------------
# SkillSync AI: application data
# ----------------------------------

students = [
    {
        "name": "Rahul",
        "branch": "CSE AIML",
        "skills": ["Python", "Machine Learning", "Pandas"],
        "interests": ["AI", "Agriculture"],
        "availability": "Evenings",
        "goal": "Build real-world AI projects"
    },
    {
        "name": "Arpit",
        "branch": "CSE",
        "skills": ["Flutter", "Dart", "UI Development"],
        "interests": ["AI", "Agriculture"],
        "availability": "Evenings",
        "goal": "Develop mobile applications"
    },
    {
        "name": "Anamika",
        "branch": "CSE",
        "skills": ["Python", "FastAPI", "PostgreSQL"],
        "interests": ["AI", "Agriculture"],
        "availability": "Weekends",
        "goal": "Build backend systems"
    }
]

projects = [
    {
        "title": "AI Crop Disease Detection",
        "owner": "Rahul",
        "description": (
            "Build an application that detects crop diseases "
            "from leaf images and provides farmer recommendations."
        ),
        "required_skills": [
            "Python", "Machine Learning", "Flutter",
            "FastAPI", "UI/UX", "Testing"
        ],
        "interests": ["AI", "Agriculture"],
        "availability": "Flexible",
        "team_size": 4
    }
]

# ----------------------------------
# Interface
# ----------------------------------

title = widgets.HTML("""
<div style="background:#302b63;color:white;padding:22px;
border-radius:12px">
<h1>🎓 SkillSync AI</h1>
<p>Find teammates. Match skills. Build amazing projects.</p>
<p>Powered by Gemma 4</p>
</div>
""")

name_box = widgets.Text(
    description="Name:",
    placeholder="Enter student name"
)
branch_box = widgets.Text(
    description="Branch:",
    placeholder="CSE AIML"
)
skills_box = widgets.Textarea(
    description="Skills:",
    placeholder="Python, Machine Learning, Flutter",
    layout=widgets.Layout(width="95%", height="65px")
)
interests_box = widgets.Text(
    description="Interests:",
    placeholder="AI, Agriculture"
)
availability_box = widgets.Dropdown(
    description="Schedule:",
    options=["Evenings", "Mornings", "Weekends", "Flexible"]
)
goal_box = widgets.Text(
    description="Goal:",
    placeholder="Build real-world projects"
)

register_button = widgets.Button(
    description="Register Student",
    button_style="success",
    icon="user-plus"
)

project_title = widgets.Text(
    description="Title:",
    placeholder="Project title",
    layout=widgets.Layout(width="95%")
)
project_owner = widgets.Dropdown(
    description="Owner:",
    options=[s["name"] for s in students]
)
project_desc = widgets.Textarea(
    description="Description:",
    placeholder="Describe the project",
    layout=widgets.Layout(width="95%", height="70px")
)
project_skills = widgets.Textarea(
    description="Skills:",
    placeholder="Python, Flutter, FastAPI",
    layout=widgets.Layout(width="95%", height="60px")
)
project_interests = widgets.Text(
    description="Interests:",
    placeholder="AI, Agriculture"
)
project_schedule = widgets.Dropdown(
    description="Schedule:",
    options=["Flexible", "Evenings", "Mornings", "Weekends"]
)
project_size = widgets.IntSlider(
    description="Team size:",
    min=2, max=10, value=4
)

publish_button = widgets.Button(
    description="Publish Project",
    button_style="success",
    icon="plus"
)

match_project = widgets.Dropdown(
    description="Project:",
    options=[p["title"] for p in projects],
    layout=widgets.Layout(width="95%")
)

match_button = widgets.Button(
    description="Ask Gemma 4 for Recommendations",
    button_style="primary",
    icon="users"
)

output = widgets.Output()

def refresh_choices():
    project_owner.options = [s["name"] for s in students]
    match_project.options = [p["title"] for p in projects]

def register_student(_):
    name = name_box.value.strip()
    branch = branch_box.value.strip()
    skills = [x.strip() for x in skills_box.value.split(",") if x.strip()]
    interests = [
        x.strip() for x in interests_box.value.split(",") if x.strip()
    ]

    with output:
        clear_output()
        if not name or not branch or not skills:
            print("Please enter name, branch and at least one skill.")
            return

        if any(s["name"].lower() == name.lower() for s in students):
            print("That student name already exists.")
            return

        students.append({
            "name": name,
            "branch": branch,
            "skills": skills,
            "interests": interests,
            "availability": availability_box.value,
            "goal": goal_box.value.strip() or "Collaborate on projects"
        })

        refresh_choices()
        print(f"Student {name} registered successfully!")

register_button.on_click(register_student)

def publish_project(_):
    with output:
        clear_output()

        title_text = project_title.value.strip()
        required = [
            x.strip() for x in project_skills.value.split(",") if x.strip()
        ]

        if not title_text or not required or not project_desc.value.strip():
            print("Enter a title, description and required skills.")
            return

        if any(p["title"].lower() == title_text.lower() for p in projects):
            print("A project with that title already exists.")
            return

        projects.append({
            "title": title_text,
            "owner": project_owner.value,
            "description": project_desc.value.strip(),
            "required_skills": required,
            "interests": [
                x.strip() for x in project_interests.value.split(",")
                if x.strip()
            ],
            "availability": project_schedule.value,
            "team_size": project_size.value
        })

        refresh_choices()
        print(f"Project '{title_text}' published successfully!")

publish_button.on_click(publish_project)

# ----------------------------------
# Gemma 4 recommendation engine
# ----------------------------------

def recommend_with_gemma(_):
    with output:
        clear_output()

        selected = next(
            (p for p in projects if p["title"] == match_project.value),
            None
        )

        if selected is None:
            print("Please select a project.")
            return

        candidates = [
            s for s in students
            if s["name"] != selected["owner"]
        ]

        if not candidates:
            print("Register at least one other student first.")
            return

        student_data = json.dumps(candidates, ensure_ascii=False)
        project_data = json.dumps(selected, ensure_ascii=False)

        prompt = f"""
You are SkillSync AI, an assistant for forming student project teams.

Analyze the project and student profiles below.

PROJECT:
{project_data}

AVAILABLE STUDENTS:
{student_data}

Your tasks:
1. Rank the most suitable students for this project.
2. Explain the matching skills and complementary skills of each candidate.
3. Suggest a role for each recommended student.
4. Identify skills missing from the combined team, considering the project
   owner's skills as well as the available candidates.
5. Consider shared interests, availability and stated goals.
6. Recommend a team that respects the project's team size.
7. If a candidate lacks a required skill, do not claim that they possess it.
8. Explain your reasoning clearly.

Return a readable report with these sections:
- Recommended team
- Candidate compatibility scores from 0 to 100
- Explanation for each recommendation
- Suggested roles
- Remaining skill gaps
- Final recommendation

Compatibility scores are estimates, not probabilities of success.
Do not invent profile information. If information is missing, say so.
"""

        print("Gemma 4 is analyzing the project and profiles...")
        print("This may take a little time.")

        try:
            messages = [
                {
                    "role": "user",
                    "content": [{"type": "text", "text": prompt}]
                }
            ]

            result = gemma(
                messages,
                return_full_text=False,
                max_new_tokens=450,
                do_sample=False
            )

            answer = result[0]["generated_text"]

            if isinstance(answer, list):
                answer = "\n".join(
                    item.get("text", "")
                    for item in answer
                    if isinstance(item, dict)
                )

            answer = str(answer).replace("<turn|>", "").strip()

            display(Markdown("## 🤖 Gemma 4 Recommendation Report"))
            display(Markdown(answer))

        except Exception as error:
            print("Recommendation failed:", str(error))
            print(
                "Check the GPU memory, model loading status, "
                "and notebook runtime."
            )

match_button.on_click(recommend_with_gemma)

# ----------------------------------
# Render application
# ----------------------------------

student_panel = widgets.VBox([
    widgets.HTML("<h2>👤 Register Student</h2>"),
    name_box, branch_box, skills_box, interests_box,
    availability_box, goal_box, register_button
])

project_panel = widgets.VBox([
    widgets.HTML("<h2>🚀 Create Project</h2>"),
    project_title, project_owner, project_desc,
    project_skills, project_interests,
    project_schedule, project_size, publish_button
])

matcher_panel = widgets.VBox([
    widgets.HTML("<h2>🧠 AI Team Matcher</h2>"),
    match_project, match_button
])

tabs = widgets.Tab(children=[
    student_panel, project_panel, matcher_panel
])
tabs.set_title(0, "Students")
tabs.set_title(1, "Projects")
tabs.set_title(2, "AI Matcher")

display(title, tabs, output)

with output:
    print("Ready! Sample students and one sample project are loaded.")
```
